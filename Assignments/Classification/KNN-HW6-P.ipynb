## 5. **Part D: Choosing $K$ without Cheating** *(15 pts)*

The number of neighbors $K$ is a **hyperparameter**:

* **Small $K$** (e.g. 1): low bias, high variance. The boundary chases every noisy point (overfitting).
* **Large $K$**: high bias, low variance. The boundary becomes too smooth (underfitting).

A tempting mistake is to pick the $K$ that maximizes **test** accuracy. That uses the test set to make a modeling decision, so the test score is no longer an honest estimate. The correct tool is **cross-validation** on the training set.
